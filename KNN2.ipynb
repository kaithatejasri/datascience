```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "weight":[2,5,7,18,4,18,15,6,19,17],
    "heights":[20,18,20,40,16,38,37,20,39,42],
    "Species":["cat","cat","cat","Lion","cat","Lion","Lion","cat","Lion","Lion"]
})
df
```

```python
X_train=df[["weight","heights"]] # Features
Y_train=df["Species"]
```

```python
model=KNeighborsClassifier(n_neighbors=4)
model.fit(X_train,Y_train)
```

```python
animal=[[24,39]]
prediction=model.predict(animal)[0]
prediction
```

```python
cats=df[df["Species"]=="cat"]
lion=df[df["Species"]=="Lion"]
```

```python
plt.figure(figsize=(8,5))
plt.scatter(cats["weight"],cats["heights"],color="blue",s=90,label="cats")
plt.scatter(lion["weight"],lion["heights"],color="red",s=90,label="Lion") 
plt.scatter(24,39,color="green",s=90,label="new Animal")
plt.title("Scatter Plot")
plt.xlabel("Weight")
plt.ylabel("Height")
plt.legend()
plt.grid(True)
plt.show()
```
