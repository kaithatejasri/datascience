```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df={"studs":["Rahul","Rohith","Hemanth","Rishitha","Raga"],
    "marks":[85,78,91,79,86],
    "age":[20,21,23,22,28]}
df=pd.DataFrame(df)
df
```

```python
plt.figure(figsize=(7,4),facecolor="pink")#it creates a canavas on which will show the graph 
plt.bar(df["studs"],df["marks"])
plt.title("Data Viz")
plt.show()
```

```python
months=["jan","feb","mar","apr","may"]
sales=[20000,23000,16000,30000,10000]
plt.figure(figsize=(7,4),facecolor="white")
plt.plot(months,sales,marker="*",markersize=10,label="sales data",lw=1,ls="--")
plt.title("Data Viz")
plt.xlabel("month")
plt.ylabel("sales")
plt.grid(True,alpha=0.2)
plt.show()
```
