```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df={"products":["Mobile","Laptop","Tablet","Watch","Computer","Ear Pods"],
"sales":[150,80,60,120,250,150]}
df=pd.DataFrame(df)
df
```

```python
plt.figure(figsize=(7,4),facecolor="white") 
plt.bar(df["products"],df["sales"])
plt.title("bar graph")
plt.show()
```

```python
years = [2020,2021,2022,2023,2024,2025,2026]
profit = [2.5,3.2,4.1,5.0,6.3,6.5,7.2]
plt.figure(figsize=(7,4),facecolor="white")
plt.plot(years,profit,marker="s",markersize=6,label="sales data",lw=1,ls="--")
plt.title("line graph")
plt.xlabel("years")
plt.ylabel("profit")
plt.grid(True,alpha=0.2)
plt.show()
```

```python
df2={"height":[150,155,160,165,170,175,180],
     "weight":[45,50,55,60,68,72,80]
}
df2=pd.DataFrame(df2)
df2
```

```python
plt.figure(figsize=(7,4),facecolor="white")
plt.scatter(df2["height"],df2["weight"],s=50,c="red",marker="o",lw=2,alpha=0.7)
plt.title("scatter plot")
plt.show()
```
