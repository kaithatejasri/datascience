```python
def parseDate(datestr):
    if pd.isna(datestr):
        return pd.NaT

    datestr = str(datestr).strip()

    formats = [
        '%Y-%m-%d',
        '%Y/%m/%d',
        '%d-%m-%Y',
        '%Y-%d-%m',
        '%d/%m/%Y'
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(datestr, format=fmt)
        except:
            continue

    return pd.NaT
```
