```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Netflix.csv")
df.columns
```

```python
df
```

```python
revenue=df["Monthly Revenue"]
revenue
```

```python
mean=sum(revenue)/len(revenue)
print(mean)
```

```python
sorted_revenue=sorted(revenue)
n=len(sorted_revenue)
if n%2==0:
    median=(sorted_revenue[n//2-1]+sorted_revenue[n//2])/2
else:
    median=sorted_revenue[n//2]
print(median)
```

```python
freq={}
for i in revenue:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1

max_freq=max(freq.values())
mode=[k for k,v in freq.items() if v==max_freq]

print(mode)
```

```python
sqrt_dev=[(x-mean)**2 for x in revenue]
variance=sum(sqrt_dev)/len(revenue)
print(variance)
```

```python
std_dev=variance**0.5
print(std_dev)
```

```python
grp=df.groupby("Subscription Type")["Monthly Revenue"].sum()
print(grp)
```

```python
agrt=df.groupby("Subscription Type")["Monthly Revenue"].agg(["sum","count","min","max","mean","median"])
print(agrt)
```
